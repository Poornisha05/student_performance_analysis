```python
performance_df = df[
    df["Pass"].isin([0, 1])
].copy()
print(performance_df["Pass"].value_counts())
```

```python
print("Total students:", len(df))
```

```python
passed = (performance_df["Pass"] == 1).sum()

print("Passed students:", passed)
```

```python
failed = (performance_df["Pass"] == 0).sum()

print("Failed students:", failed)
```

```python
passed = (performance_df["Pass"] == 1).sum()

pass_percentage = (
    passed / len(performance_df)
) * 100

print("Pass percentage:", pass_percentage)
```

```python
failed = (performance_df["Pass"] == 0).sum()

fail_percentage = (
    failed / len(performance_df)
) * 100

print("Fail percentage:", fail_percentage)
```

```python
passed_students = performance_df[
    (performance_df["Pass"] == 1) &
    (performance_df["Study_Hours"].between(0, 24))
]

print(passed_students["Study_Hours"].mean())
```

```python
print(
    performance_df.groupby("Pass")["Attendance"].mean()
)
```

```python
print(
    performance_df.groupby("Pass")["Previous_Marks"].mean()
)
```

```python
print(
    performance_df.groupby("Pass")["Backlogs"].mean()
)
```

```python
result = performance_df.groupby("Pass")[
    [
        "Study_Hours",
        "Attendance",
        "Assignments",
        "Previous_Marks",
        "Backlogs"
    ]
].mean()

print(result)
```

```python
internet_pass = (
    performance_df
    .groupby("Internet_Access")["Pass"]
    .mean() * 100
)

print(internet_pass)
```

```python
comparison = performance_df.groupby(
    "Internet_Access"
)[
    [
        "Study_Hours",
        "Attendance",
        "Previous_Marks",
        "Assignments",
        "Backlogs",
        "Pass"
    ]
].mean()

print(comparison)
```

```python
performance_df["Study_Group"] = pd.cut(
    performance_df["Study_Hours"],
    bins=[0, 2, 4, 6, float("inf")],
    labels=[
        "0–2 hours",
        "3–4 hours",
        "5–6 hours",
        "7+ hours"
    ]
)
study_pass_percentage = (
    performance_df
    .groupby("Study_Group", observed=False)["Pass"]
    .mean() * 100
)

print(study_pass_percentage)
```

```python
performance_df["Attendance_Group"] = pd.cut(
    performance_df["Attendance"],
    bins=[0, 60, 75, 90, 100],
    labels=[
        "Below 60%",
        "60–75%",
        "76–90%",
        "Above 90%"
    ]
)
attendance_pass_percentage = (
    performance_df
    .groupby(
        "Attendance_Group",
        observed=False
    )["Pass"]
    .mean() * 100
)

print(attendance_pass_percentage)
```
